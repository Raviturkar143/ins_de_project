```python
BRONZE_PATH = "/Volumes/workspace/default/insure_data/bronze"
SILVER_PATH = "/Volumes/workspace/default/insure_data/silver"
```

```python
agent_df = (
    spark.read
    .parquet(f"{BRONZE_PATH}/agent")
)
```

```python
from pyspark.sql import functions as F


silver_agent_df = (
    agent_df

    # remove duplicates
    .dropDuplicates(["agent_no"])

    # standardize text
    .withColumn(
        "first_name",
        F.upper(F.trim(F.col("first_name")))
    )

    .withColumn(
        "last_name",
        F.upper(F.trim(F.col("last_name")))
    )

    # validate status
    .filter(
        F.col("agent_status")
        .isin("ACTIVE","INACTIVE")
    )
)
```

```python
(
    silver_agent_df.write
    .mode("overwrite")
    .parquet(
        f"{SILVER_PATH}/agent"
    )
)
```

```python
print("Record Count:",
      silver_agent_df.count())

silver_agent_df.printSchema()
```
