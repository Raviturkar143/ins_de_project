```python
def get_path(layer, table_name):
    """
    Returns the S3 path for a given layer and table.
    """

    layer = layer.lower()

    if layer == "raw":
        return f"{RAW_PATH}{table_name}"

    elif layer == "bronze":
        return f"{BRONZE_PATH}{table_name}"

    elif layer == "silver":
        return f"{SILVER_PATH}{table_name}"

    elif layer == "gold":
        return f"{GOLD_PATH}{table_name}"

    else:
        raise ValueError(f"Invalid layer: {layer}")
```
